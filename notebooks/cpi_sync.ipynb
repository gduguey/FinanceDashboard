# CPI sync

Fetches the latest CPIAUCSL series from FRED and updates `data/cpi/CPIAUCSL.csv`.
Safe to re-run: only missing observations are appended.

```python
from trades.config import AppConfig
from trades.market_data import cpi as cpi_module

config = AppConfig()
```

```python
history = cpi_module.update_cpi_cache(config)
print(f"{len(history)} observations cached, {history['observation_date'].min()} to {history['observation_date'].max()}")
```
